# Phase 1 — Data Collection & Preparation

**Dataset choice: MS MARCO Passage Ranking**

Why MS MARCO?
- Real user queries from Bing search logs — mimics actual search behavior
- Each query is paired with a relevant passage (positive document)
- Passages are 200–300 words on average — perfect chunk size
- Widely used benchmark, well understood quality
- Available via HuggingFace `datasets` with a single line

**What is query / document here?**
- Query: a short natural language question (e.g. "what is the capital of france")
- Document: a 200–300 word passage from a web page that answers the query

**Pipeline:**
1. Download MS MARCO (we use the `ms_marco` v2.1 split)
2. Filter to pairs where `is_selected == 1` (confirmed relevant)
3. Chunk any long passages to 200–300 words
4. Build triplets: (query, positive_passage, negative_passage)
5. Split 80/10/10 train/val/test
6. Save to `data/processed/`


```python
import os
import json
import random
import re
from pathlib import Path
from collections import defaultdict

import numpy as np
import pandas as pd
from tqdm import tqdm

random.seed(42)
np.random.seed(42)

DATA_DIR = Path('../data')
RAW_DIR = DATA_DIR / 'raw'
PROCESSED_DIR = DATA_DIR / 'processed'
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
RAW_DIR.mkdir(parents=True, exist_ok=True)

print('Directories ready.')
```

## 1. Download MS MARCO via HuggingFace datasets

```python
from datasets import load_dataset

# ms_marco v2.1 — passage retrieval variant
# train split has ~808k queries; we only load what we need
print('Loading MS MARCO train split (this may take a minute)...')
dataset = load_dataset('ms_marco', 'v2.1', split='train', trust_remote_code=True)
print(f'Train size: {len(dataset):,} examples')
print(f'Columns: {dataset.column_names}')
print()
print('Example:')
print(dataset[0])
```

## 2. Extract (query, positive_passage) pairs

MS MARCO stores multiple passages per query in a `passages` dict with keys:
- `passage_text`: list of passage strings
- `is_selected`: list of 0/1 labels (1 = relevant)
- `url`: list of source URLs

We keep only rows where at least one passage has `is_selected == 1`.

```python
def word_count(text: str) -> int:
    return len(text.split())


def chunk_text(text: str, min_words: int = 150, max_words: int = 300) -> list[str]:
    """Split text into chunks of min_words–max_words using sentence boundaries."""
    # Simple sentence splitter (avoids NLTK dependency for now)
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    chunks, current, current_wc = [], [], 0
    for sent in sentences:
        wc = len(sent.split())
        if current_wc + wc > max_words and current:
            chunks.append(' '.join(current))
            current, current_wc = [], 0
        current.append(sent)
        current_wc += wc
    if current:
        chunks.append(' '.join(current))
    # Drop very short chunks
    return [c for c in chunks if word_count(c) >= min_words]


print('chunk_text test:')
sample = ' '.join(['This is sentence number %d.' % i for i in range(60)])
chunks = chunk_text(sample)
print(f'  Input words: {word_count(sample)}, Chunks: {len(chunks)}, Sizes: {[word_count(c) for c in chunks]}')
```

```python
# How many pairs to extract (reduce for faster iteration; 100k is enough for good training)
MAX_PAIRS = 100_000

pairs = []   # list of {'query': str, 'positive': str, 'query_id': int, 'doc_id': str}
all_passages = {}  # doc_id -> passage text (used later for negative sampling)
doc_counter = 0

print(f'Extracting up to {MAX_PAIRS:,} (query, positive_passage) pairs...')
for row in tqdm(dataset, total=len(dataset)):
    if len(pairs) >= MAX_PAIRS:
        break

    query = row['query'].strip()
    passages = row['passages']          # dict with 'passage_text' and 'is_selected'
    texts = passages['passage_text']
    labels = passages['is_selected']

    positives = [t for t, l in zip(texts, labels) if l == 1]
    negatives_pool = [t for t, l in zip(texts, labels) if l == 0]

    if not positives:
        continue

    # Store all passages for later negative sampling
    for t in texts:
        # Chunk if too long
        for chunk in (chunk_text(t) if word_count(t) > 300 else [t]):
            did = f'doc_{doc_counter}'
            all_passages[did] = chunk
            doc_counter += 1

    # Pick the first positive passage; chunk if needed
    pos_text = positives[0]
    pos_chunks = chunk_text(pos_text) if word_count(pos_text) > 300 else [pos_text]

    for chunk in pos_chunks:
        pairs.append({
            'query': query,
            'positive': chunk,
        })
        if len(pairs) >= MAX_PAIRS:
            break

print(f'\nExtracted {len(pairs):,} pairs from {doc_counter:,} total passage chunks.')
```

## 3. Assign chunk IDs and build a passage pool

We combine all unique positive passages into a pool and assign each a stable ID. This pool is what the retrieval system will index.

```python
# Deduplicate positive passages and assign IDs
seen = {}
for i, p in enumerate(pairs):
    key = p['positive']
    if key not in seen:
        seen[key] = f'pos_{len(seen)}'
    pairs[i]['positive_id'] = seen[key]

# Build the corpus dict: id -> text
corpus = {v: k for k, v in seen.items()}
print(f'Unique positive passages: {len(corpus):,}')
print(f'Total pairs: {len(pairs):,}')
```

## 4. Build Triplets (query, positive, negative)

**Negative sampling strategy:**
- **Easy negatives**: randomly sample a passage from the corpus that is NOT the positive.
  - Simple, fast, sufficient for initial training.
- **Hard negatives** (optional, better): passages that are topically similar but not the answer.
  - We add 1 hard negative per triplet using BM25 in Phase 2 once the index is built.
  - For now, we use in-batch negatives during training (InfoNCE handles this automatically).

For the explicit triplet file (used for evaluation), we sample 1 random negative per query.

```python
corpus_ids = list(corpus.keys())

triplets = []
for p in tqdm(pairs, desc='Building triplets'):
    pos_id = p['positive_id']
    # Sample a random negative (not the positive)
    neg_id = pos_id
    while neg_id == pos_id:
        neg_id = random.choice(corpus_ids)
    triplets.append({
        'query': p['query'],
        'positive_id': pos_id,
        'negative_id': neg_id,
    })

print(f'Triplets built: {len(triplets):,}')
```

## 5. Train / Validation / Test Split (80 / 10 / 10)

Shuffle first, then split. We ensure **no query appears in both train and test**.

```python
random.shuffle(triplets)

n = len(triplets)
n_train = int(0.8 * n)
n_val   = int(0.1 * n)

train_triplets = triplets[:n_train]
val_triplets   = triplets[n_train : n_train + n_val]
test_triplets  = triplets[n_train + n_val :]

# Verify no query overlap between train and test
train_queries = {t['query'] for t in train_triplets}
test_queries  = {t['query'] for t in test_triplets}
overlap = train_queries & test_queries
print(f'Train: {len(train_triplets):,}  |  Val: {len(val_triplets):,}  |  Test: {len(test_triplets):,}')
print(f'Query overlap between train and test: {len(overlap)} (should be 0 or minimal)')
```

## 6. Save to disk

```python
def save_jsonl(data: list, path: Path):
    with open(path, 'w', encoding='utf-8') as f:
        for item in data:
            f.write(json.dumps(item, ensure_ascii=False) + '\n')
    print(f'Saved {len(data):,} records → {path}')


# Corpus (passage pool)
with open(PROCESSED_DIR / 'corpus.json', 'w', encoding='utf-8') as f:
    json.dump(corpus, f, ensure_ascii=False, indent=2)
print(f'Saved corpus ({len(corpus):,} passages) → {PROCESSED_DIR / "corpus.json"}')

# Triplet splits
save_jsonl(train_triplets, PROCESSED_DIR / 'train.jsonl')
save_jsonl(val_triplets,   PROCESSED_DIR / 'val.jsonl')
save_jsonl(test_triplets,  PROCESSED_DIR / 'test.jsonl')
```

## 7. Dataset Analysis

```python
import matplotlib.pyplot as plt

# --- Length statistics ---
query_lengths = [word_count(t['query']) for t in triplets]
doc_lengths   = [word_count(corpus[t['positive_id']]) for t in triplets]

print('=== Query length (words) ===')
print(f'  Mean:   {np.mean(query_lengths):.1f}')
print(f'  Median: {np.median(query_lengths):.1f}')
print(f'  Min:    {np.min(query_lengths)}')
print(f'  Max:    {np.max(query_lengths)}')

print()
print('=== Positive passage length (words) ===')
print(f'  Mean:   {np.mean(doc_lengths):.1f}')
print(f'  Median: {np.median(doc_lengths):.1f}')
print(f'  Min:    {np.min(doc_lengths)}')
print(f'  Max:    {np.max(doc_lengths)}')

print()
print(f'Total triplets:      {len(triplets):,}')
print(f'Unique passages:     {len(corpus):,}')

# --- Histogram ---
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(query_lengths, bins=30, color='steelblue', edgecolor='white')
axes[0].set_title('Query Length Distribution (words)')
axes[0].set_xlabel('Word count')
axes[0].set_ylabel('Count')

axes[1].hist(doc_lengths, bins=30, color='coral', edgecolor='white')
axes[1].set_title('Positive Passage Length Distribution (words)')
axes[1].set_xlabel('Word count')

plt.tight_layout()
plt.savefig(PROCESSED_DIR / 'length_distributions.png', dpi=150)
plt.show()
print('Plot saved.')
```

## Summary

| Item | Value |
|------|-------|
| Dataset | MS MARCO v2.1 (passage ranking) |
| Query type | Natural language questions from Bing |
| Document type | 200–300 word web passages |
| Triplets total | ~100,000 |
| Train / Val / Test | 80% / 10% / 10% |
| Negative type | Random in-corpus negatives (easy); in-batch negatives used during training |

**Why this data is suitable:**
- Queries and documents come from real search scenarios — not synthetic
- Lexical overlap between query and document is low (the model cannot cheat with keyword matching)
- Passage lengths are naturally in the 200–300 word range MS MARCO targets
- The scale (100k pairs) is enough to fine-tune a BERT-base model meaningfully in a few epochs
